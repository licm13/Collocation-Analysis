# Exploratory Analysis (Demo)

This notebook demonstrates loading the processed demo data and plotting basic diagnostics.

示例：加载处理后的演示数据，做一些基本诊断。

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

data = np.load(Path('..')/'data'/'processed'/'et_products_demo.npz')
truth = data['truth']
products = data['products']

plt.figure()
plt.plot(truth, label='Truth')
plt.plot(products.mean(axis=0), label='EM')
plt.legend(); plt.title('Quick look'); plt.show()
```
